# 01 — Before: How Palm Oil Became Malaysia's Story

**Business question**: Before asking where the money goes in Malaysia's palm oil chain today, it's worth asking how Malaysia ended up depending on this crop at all — a West African plant that a French planter brought to Selangor in 1917. That history explains a lot about why the industry looks the way it does now: colonial-era estate ownership, a government-driven smallholder scheme built to fight rural poverty, and a decades-long rivalry with Indonesia that Malaysia eventually lost on raw scale.

## Origins: West Africa to British Malaya

The oil palm (*Elaeis guineensis*) is native to West Africa, where it was used for food and oil for centuries before European contact. It arrived in British Malaya in the **1870s**, planted purely as an ornamental in botanical gardens — nobody yet saw it as a commercial crop. **[PUBLIC]**

That changed in **1917**, when Henri Fauconnier, a French planter, established the first commercial oil palm planting at Tennamaram Estate in Selangor and built Malaysia's first palm oil mill. This is the conventional starting point the industry itself uses — MPOB marked its centenary around 2017. **[PUBLIC]**

For the next several decades, palm oil remained secondary to rubber, which was the colonial economy's dominant export crop.

## 1961: FELDA and the poverty-alleviation pivot

After independence (1957), Malaysia diversified away from over-reliance on rubber, whose price was volatile and whose smallholders were often poor. In **1961**, the Federal Land Development Authority (FELDA) began oil palm cultivation on its first land settlement scheme — 375 hectares — explicitly designed to resettle and uplift the rural landless. **[PUBLIC]**

This is the detail that matters for everything downstream in this project: Malaysia's palm oil industry was never *purely* a commercial venture. A meaningful share of it was built as deliberate government policy, which is part of why smallholders still account for roughly 39% of Malaysia's planted area today (covered in Notebook 3).

## 2007: losing the top spot to Indonesia

Malaysia was the world's largest palm oil producer for most of the 20th century. Indonesia overtook it in **2007**, driven by a much larger land base and more aggressive expansion, and has never looked back — by 2023 Indonesia's production (47.1 Mt) was more than double Malaysia's (18.6 Mt), as shown in Notebook 2. **[PUBLIC]**

That reversal is the quiet backdrop to this entire case study: Malaysia can't out-produce Indonesia on volume anymore, so the real strategic question — certification, downstream value-add, new products — is about how Malaysia competes on *something other than scale*.

## Explain It Simply

Palm oil isn't native to Malaysia — it's an African plant a French planter brought to Selangor as a business experiment in 1917. What turned it into a national industry wasn't just market demand: after independence, the government used it as a tool to lift rural smallholders out of poverty, planting it on government-organized land schemes starting in 1961. Malaysia built the world's biggest palm oil industry this way — until Indonesia, with far more land to plant on, overtook it in 2007. Everything in this project — certification, margins, the new AI-cooling-fluid bet — is really about how Malaysia tries to stay relevant in an industry it can no longer out-produce.

## Glossary

- **FFB (Fresh Fruit Bunch)**: the harvested palm fruit cluster, milled within 24–48 hours of harvest due to rapid spoilage.
- **CPO (Crude Palm Oil)**: the oil extracted from FFB flesh; the base commodity traded globally.
- **FELDA**: Federal Land Development Authority, Malaysia's government land-settlement agency, founded 1956, began oil palm schemes 1961.
- **Smallholder**: an independent or organized-scheme farmer (as opposed to a large corporate plantation) — roughly 39% of Malaysia's planted oil palm area.
- **MSPO**: Malaysian Sustainable Palm Oil certification, mandatory since 2020 (Notebook 5).

## Why Is This Happening?

**Why did Malaysia, not some other tropical country, become an early palm oil giant?** Two reinforcing factors: colonial-era plantation infrastructure and capital were already in place from the rubber industry, and after independence the government deliberately redirected that same infrastructure — plus new land schemes — toward palm oil as a poverty-alleviation and export-diversification tool. It wasn't organic market discovery; it was policy-driven from the start, which is a different story than how, say, Indonesia's oil palm sector expanded later, more through private and transmigration-linked expansion into far larger land reserves.
