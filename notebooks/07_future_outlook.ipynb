## Growth is slowing, on the record

The OECD-FAO Agricultural Outlook is the standard reference for medium-term agricultural commodity projections, jointly produced by the OECD and FAO. Comparing its two most recent editions shows a real, published downgrade:

```python
from IPython.display import Image
Image(filename='../outputs/figures/10_growth_forecast_deceleration.png')
```

The 2021-2030 edition projected 1.3%/year global palm oil supply growth; the current 2026-2035 edition projects just 0.8%/year, explicitly attributed to tightening environmental policy in importing countries constraining expansion in Indonesia and Malaysia. **[PUBLIC, OECD-FAO]** This is consistent with Notebook 2's finding that Malaysia is already near its land ceiling, and suggests Indonesia's expansion runway is narrowing too, not just Malaysia's.

Two other data points from the same outlook, both **[PUBLIC]**: biomass-based diesel already accounts for roughly 18% of global vegetable oil use and is projected to keep growing (reinforcing Notebook 6's oil-price linkage), and Indonesia + Malaysia are still projected to export more than 60% of their combined production, keeping the duopoly from Notebook 2 intact through 2035 even as growth slows.

## The next bet: palm oil as AI infrastructure

In August 2026, the Malaysian Palm Oil Board (MPOB) announced **Sawit EcoTherm**, a palm oil-based, non-conductive dielectric immersion cooling fluid designed for data centers, explicitly positioned to address two problems at once: AI servers generate 3-5x more heat than conventional servers, and Malaysia's data center boom is projected to require 876 million litres/day of water for conventional cooling, equivalent to the water needs of roughly 4 million people. Sawit EcoTherm claims 30-50% energy reduction versus conventional cooling. **[PUBLIC, MPOB via Data Center Dynamics, TechNave, The Malaysian Reserve, Aug 2026]**

**Confidence flag, stated plainly**: this is pre-commercial. There is no production volume, pricing, or customer-adoption data yet; the only concrete anchor is a planned 65MW immersion-cooled data center in Cyberjaya, backed by state-linked developer Selangor Industrial Corporation. This section is included as a genuine signal of where Malaysia is trying to take the industry next, a fifth downstream product category beyond food, oleochemicals, biodiesel, and animal feed, not as a proven revenue driver.

## What this doesn't cover, and why

## Explain It Simply

The world's official long-range forecast for palm oil just got revised down, from 1.3% growth a year to 0.8%, mainly because environmental rules are making it harder for Indonesia and Malaysia to keep expanding. At the same time, Malaysia just unveiled a genuinely surprising new bet: turning palm oil into a coolant for AI data centers, aimed at the same water-scarcity and AI-infrastructure boom reshaping the country's economy. It's not proven yet; there's no sales data, just one pilot project, but it's a real signal that Malaysia is looking for its next act beyond food and fuel.

## Glossary

- **OECD-FAO Agricultural Outlook**: the standard joint annual publication projecting 10-year agricultural commodity supply/demand.
- **Sawit EcoTherm**: MPOB's palm oil-based immersion cooling fluid for data centers, announced August 2026.
- **Immersion cooling**: a data center cooling method that submerges servers directly in a non-conductive fluid instead of using air or water-based systems.

## Why Is This Happening?

**Why would MPOB, a commodity board, get into data center cooling fluid at all?** It's a logical extension of the downstream-diversification pressure documented in Notebook 4. With upstream volume growth slowing (this notebook) and traditional downstream margins under pressure from soft demand, finding an entirely new demand category for palm-oil-derived products is a genuine strategic response, not a random pivot. Whether it succeeds commercially is unproven, but the logic connecting it to the rest of this project's findings is sound.
