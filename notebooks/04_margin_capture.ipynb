## SD Guthrie FY2025: the clearest single-company case

SD Guthrie's Group PBIT was RM3.6 billion in FY2025 (year ended December 2025). Upstream contributed 73% of that (~RM2.63bn), helped by CPO prices up 4% to RM4,254/tonne and palm kernel prices up a striking 33% to RM3,219/tonne. Downstream (SD Guthrie International, SDGI) delivered PBIT of just RM484 million, reflecting softer global demand, only partly offset by better margins in its Asia Pacific differentiated operations. Industrial Development (largely land-sale gains through a joint venture) added a further RM430 million. **[PUBLIC, company disclosure]**

By this measure, upstream earned roughly **5x** what downstream did in FY2025. Group net profit hit an all-time high of RM2.5 billion (+16% YoY), driven almost entirely by the upstream price cycle, not by the downstream diversification strategy.

## Is this an SD Guthrie story, or an industry story?

Checking against Malaysia's other major listed plantation companies (IOI Corporation, Kuala Lumpur Kepong (KLK), Genting Plantations) for the same period shows the upstream strength is sector-wide, but the downstream picture is *not* uniform, which is an important correction to a simpler "upstream good, downstream bad" story:

- **Upstream**: broadly strong across the sector. IOI's plantation segment profit surged 30% YoY, driven by a 12% rise in CPO prices and a 50% jump in palm kernel prices, an even sharper move than SD Guthrie's.
- **Downstream**: mixed, not uniformly weak. **KLK and SD Guthrie posted stronger downstream contributions**, while **Genting Plantations and IOI saw downstream profit declines**, pressured by softer refining margins and continued weakness in oleochemical demand against elevated feedstock costs. **[DERIVED, synthesized from sector coverage: The Edge Malaysia, BusinessToday, Kenanga sector research]**

So the honest finding is narrower than "downstream doesn't pay": **upstream is riding a strong price cycle across the board, while downstream performance now depends on company-specific execution and product mix**. SDGI's specific weakness (RM484m) reflects SD Guthrie's own downstream exposure, not an industry-wide downstream failure.

## Explain It Simply

In FY2025, Malaysia's biggest plantation company made about five times more money from simply growing and selling raw palm fruit (upstream) than from all its refining, oleochemicals, and branded-product business combined (downstream), even though the whole reason Malaysia pushed companies downstream was to escape being just a raw-commodity seller. But that's not true everywhere: two of the four major players actually grew their downstream profits this year, while two saw them shrink. So the honest answer isn't "downstream investment was a mistake," it's "a strong commodity price cycle is currently doing more for profits than the downstream strategy is, and not every company is executing downstream equally well."

## Glossary

- **PBIT**: Profit Before Interest and Tax, the standard segment-profitability measure used in these companies' financial disclosures.
- **Upstream**: plantation cultivation and milling (FFB to CPO/PK).
- **Downstream**: refining, fractionation, oleochemicals, biodiesel, and branded consumer products.
- **SDGI**: SD Guthrie International, the group's downstream operating arm.

## Why Is This Happening?

**Why would downstream margins shrink while the underlying commodity (palm oil) gets more expensive?** Downstream businesses buy CPO as a feedstock and sell refined/processed products. When feedstock (CPO) prices rise faster than what refiners can pass on to customers in a soft-demand market, refining and oleochemical margins get squeezed from both sides: higher input costs, weaker pricing power. That's exactly what sector commentary describes for FY2025: "softer refining margins and continued pressure in oleochemicals from weak demand and elevated feedstock costs." It's a genuine structural tension in vertically-integrated plantation groups, not a one-off.
