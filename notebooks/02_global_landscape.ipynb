# 02: Global Landscape, Who Grows Palm Oil, and Why Just Two Countries

**Business question**: Notebook 1 showed palm oil arrived in Malaysia by historical accident and grew through deliberate policy. This notebook establishes the scale of what that produced: as of 2023, Indonesia and Malaysia together account for 83% of global palm oil production. Before comparing the two head-to-head (Notebook 3), it's worth seeing just how lopsided the global picture is, and how Malaysia lost its lead.

## Data

Source: FAOSTAT Production, Crops and Livestock domain (Item: Palm oil, Element: Production), bulk download, filtered to real countries only (regional/income-group aggregates such as "Least Developed Countries" and "Melanesia" excluded via FAOSTAT's own area-code convention, where aggregate groupings carry Area Codes ≥ 1000). **[PUBLIC]**. Latest complete year at time of writing is 2023; 2024 data was incomplete in the bulk file.

```python
import pandas as pd
from IPython.display import Image

prod2023 = pd.read_csv('../data/processed/global_palm_oil_production_2023.csv')
world_total = prod2023['Value'].sum()
top2 = prod2023[prod2023['Area'].isin(['Indonesia', 'Malaysia'])]['Value'].sum()
print(f"Malaysia + Indonesia share of tracked countries' production: {top2 / world_total:.1%}")
prod2023
```

No other country comes close. Thailand, the third-largest producer, grows less than a tenth of what Malaysia does. This is a genuine duopoly, not just a "top two."

## How the duopoly formed: Indonesia's 2007 overtake

Malaysia was the world's largest producer for most of the 20th century, on the back of the FELDA-driven expansion covered in Notebook 1. Indonesia overtook it in 2007 and has pulled steadily further ahead since, driven by a much larger available land base for expansion.

```python
trend = pd.read_csv('../data/processed/malaysia_indonesia_production_timeseries.csv', index_col='Year')
crossover_year = trend[trend['Indonesia'] > trend['Malaysia']].index.min()
latest = trend.iloc[-1]
print(f"Indonesia first overtook Malaysia in: {crossover_year}")
print(f"Latest year (Malaysia vs Indonesia, tonnes): {latest['Malaysia']:,.0f} vs {latest['Indonesia']:,.0f}")
trend.tail()
```

Two things stand out. First, Malaysia's production has been essentially flat since roughly 2009, hovering between 17 and 20 million tonnes for 15 years, while Indonesia's nearly tripled over the same period. Second, this isn't a story of Malaysia declining; it's a story of Malaysia hitting a land ceiling while Indonesia kept expanding. That structural difference is the subject of Notebook 3.

## Explain It Simply

Two countries, Indonesia and Malaysia, grow more than 8 out of every 10 tonnes of palm oil produced on Earth. Malaysia used to be the bigger of the two, but Indonesia overtook it in 2007 simply because Indonesia has far more land available to plant on. Malaysia's production has barely moved in 15 years, not because something went wrong, but because it's running out of room to grow the old way. That's the real reason the rest of this project looks at margins, certification, and new products instead of just "grow more": for Malaysia, growing more isn't really on the table anymore.

## Glossary

- **Duopoly**: a market dominated by two suppliers, here, Indonesia and Malaysia at a combined 83% of global production.
- **Land ceiling**: the practical limit on planted area a country can add, driven by available arable/forested land, not demand.

## Why Is This Happening?

**Why can't Malaysia just plant more palm oil to compete with Indonesia on scale?** Malaysia's total land area is about a quarter of Indonesia's, and much of what's suitable for oil palm is already planted; the country has been at or near its practical planted-area ceiling since the mid-2000s. Indonesia, by contrast, still had large forested/peatland frontier areas (Sumatra, Kalimantan) to convert during the 2007-2020 expansion, which is also the period that drew the deforestation scrutiny behind the EU's regulatory response (Notebook 5). Malaysia's strategic options are therefore structurally different from Indonesia's: it can't out-grow the problem, so it has to out-earn it.
