# 06: Price & Demand, What the 2026 Oil Crisis Changed

**Business question**: In 2026, the Iran war and the closure of the Strait of Hormuz triggered what the IEA called the largest oil supply disruption in the history of the global oil market. Palm oil is a biodiesel feedstock, which links its price to crude oil through fuel-substitution economics, the "POGO spread" (palm oil vs. gas oil). Has this relationship shown up historically, and what did the 2026 shock do to it?

**Data note**: an earlier draft of this notebook used a third-party site's redistributed MPOB price series. On review, that site's license explicitly prohibits redistributing its data publicly, so it was removed and replaced with the World Bank's Commodity Markets ("Pink Sheet") data, which is CC BY 4.0 licensed and includes both Brent crude and palm oil price series from the same authoritative source. The tradeoff: the World Bank series currently runs only through December 2025 (publication lag), so the 2026 crisis period itself is discussed narratively below, citing published market reporting, rather than plotted from a redistributed dataset.

## The historical relationship: real, and visible well before 2026

```python
import pandas as pd
from IPython.display import Image

wb = pd.read_csv('../data/processed/wb_brent_vs_palmoil_monthly.csv', parse_dates=['month'])
corr = wb['Crude oil, Brent'].corr(wb['Palm oil'])
print(f"Correlation between monthly Brent crude and palm oil prices ({wb['month'].min():%Y-%m} to {wb['month'].max():%Y-%m}): r={corr:.2f}")
wb.tail()
```

```python
Image(filename='../outputs/figures/08_brent_vs_palmoil_2018_2025.png')
```

The correlation shown above is a real, moderate-to-strong positive relationship. It's most visible in 2022, when both series spiked together during the Russia-Ukraine war energy shock, then fell back in tandem through 2023. **[PUBLIC, World Bank Commodity Markets, CC BY 4.0]**

This is the mechanism the 2026 crisis was expected to trigger again: when crude oil (and gas oil/diesel) gets more expensive, palm-oil-based biodiesel becomes a relatively more attractive fuel substitute, pulling palm oil demand and price up with it (the "POGO spread").

## What happened in the 2026 crisis

Brent crude rose from roughly $64 to $104/barrel over about five months in early 2026 following the Iran war and Strait of Hormuz closure. Trade press (OFI Magazine) reported this compressed the POGO spread to a 41-month low of roughly -$292/tonne, meaning palm oil became *cheaper relative to gas oil* than it had been in over three years, which on the standard mechanism should make biodiesel substitution more attractive and pull palm oil demand upward. **[PUBLIC, OFI Magazine reporting]** By September 2026, Brent/WTI had settled above $100, with the Strait of Hormuz near-standstill (vessel traffic down ~95%) and the IEA calling it the largest oil supply disruption on record. **[PUBLIC, IEA, CNBC, Wikipedia]**

This notebook does not have a redistributable, verified palm oil price series covering the 2026 window itself to plot alongside this; that's a genuine, disclosed data gap, not a claim this project can fully back with its own chart. The historical relationship computed above is the solid, demonstrated finding; the 2026-specific reaction is reported by trade press but not independently re-verified here.

India alone buys roughly 22% of globally traded palm oil, with China, the EU, Pakistan, and the US rounding out the top five buyers at a combined ~55%. **[PUBLIC, compiled from customs trade data via industry trade analyses]** This concentration means a demand shock in any one of these markets has an outsized effect on global palm oil prices, and by extension on Malaysia's export revenue, a structural vulnerability layered on top of the oil-price linkage above.

## Explain It Simply

Palm oil price has moved with crude oil price for years, because palm oil can be turned into diesel fuel: when oil gets expensive, biodiesel demand (and palm oil demand) tends to rise too. That relationship showed up clearly during the 2022 energy shock. In 2026, a war involving Iran shut down a key oil shipping route and sent oil prices spiking again, and trade press reports the same mechanism kicked in, but we couldn't independently verify Malaysia-specific palm oil price data for that exact period without using a data source whose terms don't allow us to republish it, so we're reporting that part honestly as "trade press says X," not as our own confirmed chart. On top of the oil-price link, just five countries buy more than half of all palm oil traded globally, so demand shocks concentrate fast.

## Glossary

- **POGO spread**: Palm Oil to Gas Oil spread, the price difference traders watch to judge whether biodiesel substitution is economically attractive.
- **Pink Sheet**: the World Bank's Commodity Markets monthly price data publication, the source used here.

## Why Is This Happening?

**Why isn't the correlation stronger, given the POGO mechanism is well-established?** Palm oil price is also driven by its own independent supply cycle: weather affecting FFB yields (Notebook 3), competing vegetable oil supply, and importing-country stock/policy decisions, all moving independently of crude oil. A moderate, not perfect, correlation is what a real but partial demand-side linkage layered on an agricultural commodity's own supply cycle should look like.
