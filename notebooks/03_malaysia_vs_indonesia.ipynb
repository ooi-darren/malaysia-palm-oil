# 03 — Malaysia vs. Indonesia: The Head-to-Head

**Business question**: Notebook 2 showed Indonesia pulling ahead of Malaysia in total volume since 2007. This notebook asks the sharper question: is that a story of Malaysia falling behind, or of Malaysia hitting a genuinely different constraint? The answer matters for anyone deciding whether Malaysia is a declining producer to avoid, or a maturing, more efficient one worth backing on different terms.

## Land vs. efficiency

```python
from IPython.display import Image
Image(filename='../outputs/figures/04_area_vs_yield_my_id.png')
```

Indonesia's harvested area nearly tripled between 2010 and 2024 (4.9M to 14.3M hectares), while Malaysia's has been essentially flat at ~5.1M hectares since 2010 -- consistent with the land-ceiling explanation from Notebook 2. But on a per-hectare basis, Malaysia's oil palm fruit yield has generally run *higher* than Indonesia's over the past decade (roughly 18,500-19,800 kg/ha vs. Indonesia's 16,900-18,500 kg/ha in recent years). **[PUBLIC, FAOSTAT]**

This is the notebook's central finding: **Malaysia isn't losing on efficiency, it's losing on land**. Indonesia's volume lead comes from planting on roughly 2.8x the area, not from getting more out of each hectare.

## Smallholder structure: more similar than expected

Malaysia's smallholders hold about 38.8% of national planted area (MPOB); Indonesia's hold 42.9% (BPS, Statistics Indonesia, 2024). **[PUBLIC]** These are close enough that neither country has a structural smallholder-exposure advantage over the other heading into stricter EU traceability requirements (EUDR, covered in Notebook 5) -- both have a very large base of small, harder-to-trace farms to bring into compliance.

## Aging trees: a shared problem, not a uniquely Malaysian one

It's tempting to assume Malaysia's flat output reflects an aging, past-peak tree stock needing replanting. That's real, but research on Indonesia's smallholder sector shows the same aging/replanting problem there -- Indonesia's own government replanting subsidy has been reported as insufficient to get smallholders to replant aging palms at the pace needed. **[PUBLIC]** This isn't a Malaysia-specific weakness; it's a maturing-industry problem both countries share, just arriving on different timelines given Indonesia's more recent expansion.

## Explain It Simply

Indonesia produces more than twice as much palm oil as Malaysia, but not because it's better at farming -- it's because it has nearly three times as much land planted. Hectare for hectare, Malaysia actually gets slightly more fruit out of its land than Indonesia does. Both countries also have a similar share of small, independent farmers (not big companies) growing the crop, and both face the same looming problem: a lot of their oil palm trees are getting old and need replanting, which is expensive and cuts output for a few years. So Malaysia's story isn't decline -- it's a country that ran out of room to expand the old way, which is exactly why the rest of this project looks at margins, certification, and new products rather than more planting.

## Glossary

- **Yield (kg/ha)**: fresh fruit bunch output per hectare of harvested area -- the standard efficiency measure for oil palm.
- **Replanting**: removing aging, lower-yielding palms (typically past 20-25 years) and replanting -- a multi-year capital investment with an output gap while new trees mature.

## Why Is This Happening?

**Why does Malaysia out-yield Indonesia despite having a smaller, older industry?** A plausible driver (not confirmed at the level of a formal causal study here) is that Malaysia's plantation sector is more consolidated among larger, better-resourced corporate estates -- the companies studied in Notebook 4 -- which can afford better fertilizer regimes, drainage, and replanting discipline than Indonesia's larger and more fragmented smallholder base spread across a much bigger, more geographically dispersed planted area. This is a hypothesis the data here supports directionally but doesn't prove -- a genuine limitation of this notebook.
